# 08 — Look at a field or a pixel: NDVI curve, radar curves, Sentinel-2 chips

Two ways in, both from what you see in QGIS:

* **a field**: every field in `fields_all_aois.gpkg` (or `fields/aoi<N>_fields_monsoon2026.gpkg`) has a
  unique **`field_id`** such as `aoi116_000002` (identify the field in QGIS and copy it). The curves are
  averaged over all the field's pixels and the field outline is drawn on the chips.
* **a pixel**: load the AOI's pixel-index raster `processed/aoi<N>/monsoon2026/grid/pixel_index.tif`
  in QGIS; its **band 1** value under the cursor is the **pixel id**. Give the AOI number and pixel id;
  a small block of pixels around it (1x1, 3x3 or 5x5) is shown.

What you see: NDVI (filled dots = clear observation, hollow = hazy, line = the fitted curve the rule
reads, with the trough and climb it found), VV (solid) and VH (dotted) of every radar track, and every
clear 5-3-2 chip (red edge, green, blue). The class in every map version and the field's label and
confidence are printed. Pictures are also saved in `processed/_batch/s2_2026/qgis_review/inspect/`.
Each look takes about a minute (the AOI's images are read once).

Code: `src/sar_pipeline/qgis_review.py` (`inspect_field`, `inspect_pixel`),
`src/sar_pipeline/analysis/water_investigation.py` (`group_sheet`).

**Rule label on the plot (live):** the title shows the class the CURRENT rules give this pixel / field
(`analysis.curve_rules`, computed when you run the cell, so it changes as soon as a rule changes) and your own label
if you gave one (`analysis.curve_labels`). The box on the right lists the measured features the rule read: where
VH / VV are now in the pixel's own season range, their recent change, NDVI low/peak, share of the rise left,
green-up days, days since the peak and whether water stood while the crop was small.
